## Exercice 1 : Création et restructuration d'un tableau (1D → 2D)

## Exercice 2 : Extraction par masque booléen (1D)

**Objectif :**
Soit un vecteur 1D de 15 nombres aléatoires entre 0 et 50. Extrayez tous les éléments strictement supérieurs à 25 **ET** pairs.

```python

# VOTRE CODE ICI

a = np.random.randint(0,50,15)
print(a)

print((a>25) & (a%2 ==0)) # return array of boolean values

# Extract
result = a[(a>25) & (a%2 ==0)]

print(result)




```

## Exercice 3 : Inversion et inversion par colonnes (1D et 2D)

**Objectif :**
1. Inversez l'ordre des éléments d'un vecteur 1D de taille 10.
2. Soit une matrice $4 \times 4$, inversez l'ordre de ses colonnes (la dernière colonne devient la première).

```python
# VOTRE CODE ICI
a = np.random.randint(0,100,10)

print(a)

# Inverse
inverse = a[::-1]
print(inverse)


# Inverse Colonnes
a = np.random.randint(0,100,16).reshape(4,4)
print(a)
inverse_colonnes = a[:,::-1]

print(inverse_colonnes)



```

## Exercice 4 : Statistique par axe (2D)

**Objectif :**
Soit une matrice $5 \times 3$ d'entiers aléatoires entre 1 et 100.
1. Calculez la moyenne globale de la matrice.
2. Calculez la somme des éléments pour chaque **colonne**.
3. Obtenez la valeur maximale pour chaque **ligne**.

```python
# VOTRE CODE ICI

matrice = np.random.randint(1,100,15).reshape(5,3)
print(matrice)

moyen = matrice.mean()
print(f"moyen: {moyen}")
somme_colonne = np.sum(matrice,axis=0)
print(f"somme_colonnes:\n {somme_colonne}")

max_lines = np.max(matrice,axis=1)
print(f"max_lines:\n {max_lines}")

```

## Exercice 5 : Remplacement des valeurs manquantes (1D / 2D)

**Objectif :**
Soit le tableau : `arr = np.array([1.0, 2.0, np.nan, 4.0, np.nan, 6.0])`.
Remplacez toutes les valeurs `NaN` par la moyenne des valeurs non manquantes du tableau.

```python
# VOTRE CODE ICI
arr = np.array([1.0,2.0,np.nan,4.0,np.nan,6.0])
print(f"Original:\n{arr}")
# Moyen
moyenne = np.nanmean(arr)
print(f"moyen: {moyenne}")

print(np.isnan(arr))
# Remplacement
arr[np.isnan(arr)] = moyenne

print(f"Result:\n{arr}")


```

## Exercice 6 : Concaténation et empilement (2D)

**Objectif :**
Soit deux matrices $A$ et $B$ de dimensions $3 \times 2$.
1. Empilez-les verticalement pour obtenir une matrice $6 \times 2$.
2. Empilez-les horizontalement pour obtenir une matrice $3 \times 4$.

```python
# VOTRE CODE ICI

A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

C_vertical = np.vstack((A,B))

C_horizontal = np.hstack((A,B))

print(C_vertical)
print(C_horizontal)


```

## Exercice 7 : Recherche d'indices spécifiques (`np.where`) (1D)

**Objectif :**
Soit un vecteur 1D de 12 valeurs réelles. Récupérez les **indices** (et non les valeurs) de tous les éléments compris entre 10 et 20 inclus.

## Exercice 8 : Calcul de la diagonale et trace (2D)

**Objectif :**
Générez une matrice carrée $4 \times 4$ d'entiers.
1. Extrayez les éléments de la diagonale principale sous forme de vecteur 1D.
2. Calculez la somme de ces éléments (la trace) en utilisant la fonction NumPy dédiée.

## Exercice 9 : Masquage conditionnel & Remplacement rapide (1D)

**Objectif :**
Créez un vecteur 1D de 20 entiers aléatoires entre -10 et 10. Sans utiliser de boucle for ni d'instructions if, remplacez tous les nombres négatifs impairs par leur carré, et tous les nombres positifs pairs par 0.

```python
# VOTRE CODE ICI

a = np.random.randint(-10,10,20)

print(f"Original:\n {a}")

# negative impaire -> leur carre

a = np.where(((a<0 ) & (a % 2 != 0)), a**2,a )
print(f"negative impaire -> leur carre:\n {a}")

# positive to 0
a = np.where((a>0),0,a)

print(f"positive to 0:\n {a}")





```

## Exercice 10 : Extraction par glissement / Fenêtres glissantes (1D)

**Objectif :**
Soit un vecteur 1D $V = [1, 2, 3, 4, 5, 6, 7, 8]$. Générez une matrice 2D représentant toutes les sous-fenêtres glissantes de taille 3 avec un pas (stride) de 1.Résultat attendu :$$\begin{bmatrix} 1 & 2 & 3 \\ 2 & 3 & 4 \\ 3 & 4 & 5 \\ \dots & \dots & \dots \end{bmatrix}$$

```python
# VOTRE CODE ICI
a = np.arange(1,9)
print("Original:\n",a)
fenetre = np.lib.stride_tricks.sliding_window_view(a,3)
print("fenetre:\n",fenetre)

# solution NR

v = np.arange(1,9)
window_size = 3
result = []
for n in range(len(v) - window_size +1):
    result.append(v[n:n+window_size])

result = np.array(result)
print("fenetre_for_loop:\n",result)



```

## Exercice 11 : Extraction des k plus grands éléments par ligne (2D)

**Objectif :**
Soit une matrice 2D de dimensions $6 \times 8$ remplie d'entiers aléatoires. Extrayez, pour chaque ligne, les 3 plus grandes valeurs et leurs indices respectifs, triés par ordre décroissant.

```python
# VOTRE CODE ICI

a = np.random.randint(0,100,(6,8))

print("Original:\n",a)
indices_tries = np.argsort(a,axis=1)
print("Indice:\n",indices_tries)

indices_3 = indices_tries[:,-3:]
print("Indice_3:\n",indices_3)

# Inverse 
indices_3 = indices_3[:,::-1]
print("Indice_3_inverse:\n",indices_3)


# valeurs
valeurs_3 = np.take_along_axis(a,indices_3,axis=1)
print("valeurs final:\n",valeurs_3)



```
