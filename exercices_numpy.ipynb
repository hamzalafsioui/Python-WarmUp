## Exercice 2 : Extraction par masque booléen (1D)

**Objectif :**
Soit un vecteur 1D de 15 nombres aléatoires entre 0 et 50. Extrayez tous les éléments strictement supérieurs à 25 **ET** pairs.

```python

# VOTRE CODE ICI

a = np.random.randint(0,50,15)
print(a)

print((a>25) & (a%2 ==0)) # return array of boolean values

# Extract
result = a[(a>25) & (a%2 ==0)]

print(result)




```

## Exercice 3 : Inversion et inversion par colonnes (1D et 2D)

**Objectif :**
1. Inversez l'ordre des éléments d'un vecteur 1D de taille 10.
2. Soit une matrice $4 \times 4$, inversez l'ordre de ses colonnes (la dernière colonne devient la première).

```python
# VOTRE CODE ICI
a = np.random.randint(0,100,10)

print(a)

# Inverse
inverse = a[::-1]
print(inverse)


# Inverse Colonnes
a = np.random.randint(0,100,16).reshape(4,4)
print(a)
inverse_colonnes = a[:,::-1]

print(inverse_colonnes)



```

## Exercice 4 : Statistique par axe (2D)

**Objectif :**
Soit une matrice $5 \times 3$ d'entiers aléatoires entre 1 et 100.
1. Calculez la moyenne globale de la matrice.
2. Calculez la somme des éléments pour chaque **colonne**.
3. Obtenez la valeur maximale pour chaque **ligne**.

```python
# VOTRE CODE ICI

matrice = np.random.randint(1,100,15).reshape(5,3)
print(matrice)

moyen = matrice.mean()
print(f"moyen: {moyen}")
somme_colonne = np.sum(matrice,axis=0)
print(f"somme_colonnes:\n {somme_colonne}")

max_lines = np.max(matrice,axis=1)
print(f"max_lines:\n {max_lines}")

```
