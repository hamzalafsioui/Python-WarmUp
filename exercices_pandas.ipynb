## Exercice 1 : Création de DataFrame et inspection

**Objectif :**
Créez un DataFrame `df1` à partir du dictionnaire fourni. Affichez ses 3 premières lignes, ses types de colonnes et ses statistiques descriptives (`describe()`).

```python
data1 = {
    'Nom': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, 40, 22, 28],
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Paris', 'Lyon', 'Lille'],
    'Salaire': [35000, 42000, 50000, 48000, 29000, 38000]
}

# VOTRE CODE ICI

df1 = pd.DataFrame(data1)


print(df1)
print("\nFirst 3 lines\n")
print(df1.head(3))

print("\nType of colonnes\n")
print(df1.dtypes)

print("\nStatics descriptives\n")
print(df1.describe())
```

## Exercice 2 : Filtrage conditionnel (`loc` & masque)

**Objectif :**
Extrayez les personnes âgées de **plus de 25 ans** vivant à **Paris** ou **Lyon**, et ne conservez que les colonnes `'Nom'` et `'Salaire'`.

```python
# VOTRE CODE ICI

masque = (df1['Age'] > 25) & (df1['Ville'].isin(['Paris','Lyon']))


result = df1.loc[masque,["Nom","Salaire"]]

print(masque)
print(result)
```

## Exercice 3 : Gestion des valeurs manquantes

**Objectif :**
1. Identifiez le nombre de valeurs `NaN` par colonne.
2. Remplacez les valeurs manquantes de la colonne `'Score'` par sa médiane.
3. Supprimez les lignes où `'Categorie'` est manquante.

```python
df3 = pd.DataFrame({
    'Score': [85.0, np.nan, 92.0, 78.0, np.nan, 88.0],
    'Categorie': ['A', 'B', np.nan, 'A', 'B', 'B'],
    'Ventes': [100, 150, 120, 90, 110, 130]
})
print("DataFrame original :\n", df3)

# VOTRE CODE ICI

print(df3.isna().sum())

# 2  remplacement score nan
median = df3['Score'].median()
print("median: ",median)
df3['Score'] =  df3['Score'].fillna(median)

# delete manque elements in category
df3 = df3.dropna(subset='Categorie')

print("DataFrame original :\n", df3)



```

## Exercice 4 : Aggrégation et GroupBy

**Objectif :**
À partir du DataFrame `df1`, calculez par **Ville** :
1. Le salaire moyen.
2. L'âge maximum.
3. Le nombre total d'employés.
*(Indice : utilisez `.groupby()` avec `.agg()`)*

```python
# VOTRE CODE ICI
result = df1.groupby('Ville').agg(
    salaire_moyen = ('Salaire','mean'),
    age_max = ('Age','max'),
    nombre_employes = ('Nom','count')
)

print(result)
```

## Exercice 5 : Création de nouvelles colonnes et fonctions personnalisées

**Objectif :**
Ajoutez une colonne `'Tranche_Age'` à `df1` :
- `'Jeune'` si l'âge est $< 30$
- `'Senior'` si l'âge est $\ge 30$

*(Utilisez `np.where()` ou `.apply()`)*

```python
# VOTRE CODE ICI

df1['Tranche_Age'] = np.where(df1['Age'] < 30,'Jeune','Senior')

print(df1)
```

## Exercice 6 : Fusion de DataFrames (`merge` / `join`)

**Objectif :**
Réalisez une jointure à gauche (*Left Join*) entre le DataFrame des employés `df1` et le DataFrame des départements `df_deps` sur la colonne `'Ville'`.

```python
df_deps = pd.DataFrame({
    'Ville': ['Paris', 'Lyon', 'Marseille', 'Toulouse'],
    'Departement': ['RH', 'IT', 'Ventes', 'Marketing']
})

# VOTRE CODE ICI

result = pd.merge(
    df1,
    df_deps,
    on='Ville',
    how='left'
)

print(result)
```

## Exercice 7 : Tri et réindexation

**Objectif :**
Triez `df1` par **Ville** (ordre alphabétique croissant), puis par **Salaire** (ordre décroissant). Réinitialisez ensuite l'index sans conserver l'ancien index.

```python
# VOTRE CODE ICI

df1 = df1.sort_values(
    by=["Ville","Salaire"],
    ascending=[True,False]
)

df1 = df1.reset_index(drop=True)

print(df1)

```
