2. <span style="font-weight: bold; color: #e56922ff;">Travail à faire</span>:</br>
Crée une classe CompteBancaire avec les caractéristiques suivantes :
1. Attributs :<br>
* nom_proprietaire (chaîne)<br>
* solde (float, initialisé à 0.0 par défaut)<br>
2. Méthodes :
*  __init__() : initialise le compte avec le nom du propriétaire et un solde optionnel.
* deposer(montant) : ajoute le montant au solde.
* retirer(montant) : retire le montant du solde si suffisant, sinon affiche un message d’erreur.
* afficher_solde() : affiche le nom du propriétaire et le solde actuel.

```python
#  Écrivez du code Python ici

class CompteBancaire:
    def __init__(self,nom_proprietaire:str,sold:float = 0.0):
        self.nom_proprietaire = nom_proprietaire
        self.sold = sold

    def deposit(self,montant):
        self.sold += montant

    def withdraw(self,montant):
        if(self.sold < montant):
            raise ValueError( 
                "Your Bank Amount does not have enough money to withdraw it !!!"
                )
        self.sold -= montant

    def display_Info(self):
        print(f"Name: {self.nom_proprietaire} | Sold: {self.sold}")


person1 = CompteBancaire("Hamza",1000)
person1.deposit(250)
person1.display_Info()
person1.withdraw(800)
person1.display_Info()

try:
    person1.withdraw(800)
except Exception as e:
    print(e)



```

```python
#  Écrivez du code Python ici

from abc import abstractmethod, ABC
class Person(ABC):
    def __init__(self,lname,fname,age):
        self.lname = lname
        self.fname = fname
        self.age = age


    @abstractmethod
    def display_info(self):
        pass

class Student(Person):
    def __init__(self, lname, fname, age,matricule,notes:list[float]):
        super().__init__(lname, fname, age)
        self.matricule = matricule
        self.notes = notes

    def ajoute_note(self,note:float):
        self.notes.append(note)

    def moyen(self)->float:
        m = sum(self.notes)/len(self.notes)
        return m

    def display_info(self):
        print(f"fName:{self.fname} | lName:{self.lname} | Age:{self.age} | notes:{self.notes}")


class Teacher(Person):
    def __init__(self, lname, fname, age,specialite,salaire):
        super().__init__(lname, fname, age)
        self.specialite = specialite
        self.salaire = salaire # calls the setter (self.salaire = salaire)
        # self._salaire = salaire // the setter is not called during initialization

    def display_info(self):
        print(f"fName:{self.fname} | lName:{self.lname} | Age:{self.age} | specialite:{self.specialite} | salaire:{self.salaire}")


    @property
    def salaire(self):
        return self._salaire

    @salaire.setter
    def salaire(self,new_salaire):
        if(new_salaire<0):
            raise ValueError(f"Salaire must be positive [Student Name: {self.fname}_{self.lname}]")
        else:
            self._salaire = new_salaire


class Scholl:
    def __init__(self, nom:str , list_students:list[Student] , list_teachers:list[Teacher]) :
        self.nom = nom
        self.list_students = list_students
        self.list_teachers = list_teachers

    def ajouter_student(self, etudiant:Student) :
        self.list_students.append(etudiant)
    
    def ajouter_teacher(self, enseignent:Teacher) :
        self.list_teachers.append(enseignent)


    def display_all(self):
        print("So List of Students : \n")
        for student in self.list_students :
            student.display_info()
            print("\n")
        
        print("\nSo List of Teachers : \n")
        for teacher in self.list_teachers :
            teacher.display_info()
            print("\n")




student_1 = Student("Ali", "Lotfi", 20, 101, [15, 18, 12])
student_2 = Student("Sara", "Hassan", 21, 102, [16, 14, 19])
student_3 = Student("Yassine", "Khalid", 22, 103, [10, 12, 14])
student_4 = Student("Meryem", "Nabil", 20, 104, [18, 17, 20])
student_5 = Student("Hicham", "Rachid", 23, 105, [13, 15, 12])


teacher1 = Teacher("Driss", "Ahmed", 40, "Mathématiques", 5000)
teacher2 = Teacher("Fatima", "Laila", 35, "Physique", 4800)
teacher3 = Teacher("Karim", "Omar", 45, "Informatique", 5500)
teacher4 = Teacher("Imane", "Said", 38, "Chimie", 4700)
teacher5 = Teacher("Rachid", "Sofia", 50, "Biologie", 6000)


scholl = Scholl("Ibn_Salam", [student_1,student_4,student_5], [teacher2,teacher3])

scholl.display_all()

# update salaire
print("="*30)
try:
    teacher3.salaire = -1202
except Exception as e:
    print(e)

print("="*30)


scholl.display_all()

    
```

```python
from datetime import date
from abc import ABC,abstractmethod

class Employe(ABC):
    def __init__(self,nom,prenom,matricule,anne_de_naissance):
        self.nom = nom
        self.prenom = prenom
        self.anne_de_naissance = anne_de_naissance
        self.matricule = matricule

    def __str__(self):
        return f"Nom:{self.nom} | Prenom:{self.prenom} | Matricule:{self.matricule} | Anne_de_Naissance:{self.anne_de_naissance}"

    @abstractmethod
    def get_salaire(self):
        pass


class Ouvrier(Employe):
    SMIG = 2500

    def __init__(self, nom, prenom, matricule, anne_de_naissance,anne_entre):
        super().__init__(nom, prenom, matricule, anne_de_naissance)
        self.anne_entre = anne_entre

    
    def get_salaire(self):
        anne_actuel = date.today().year
        anciente = anne_actuel - self.anne_entre

        salaire = self.SMIG + (anciente) * 100
        salaire_max = self.SMIG *2

        return min(salaire,salaire_max)

    def __str__(self):
        return super().__str__() + f"anne_de_entree:{self.anne_entre} | salaire: {self.get_salaire()} DH"


class Cadre(Employe):
    def __init__(self, nom, prenom, matricule, anne_de_naissance,indice):
        super().__init__(nom, prenom, matricule, anne_de_naissance)
        self.indice = indice

    def get_salaire(self):
        salaires = {
            1:13000,
            2:15000,
            3:17000
        }
        if self.indice not in salaires:
            raise ValueError(f"Indice Must Be 1,2 or 3 | cadre_name:{self.nom}_{self.prenom}")
        
        return salaires[self.indice]

    def __str__(self):
        return super().__str__() + f"| Indice: {self.indice} | Salaire menduel: {self.get_salaire()} DH"

class Patron(Employe):
    chiffre_affaire = 0 
    def __init__( self, nom, matricule, prenom, annee_naissance ): 
        super().__init__( nom, matricule, prenom, annee_naissance )

    def get_salaire(self): 
        return Patron.chiffre_affaire * 0.1

    def __str__(self):
        return super().__str__() + f" | Chiffre affaire: {Patron.chiffre_affaire} | salaire_mensuel: {self.get_salaire()}"



ouvrier_1 = Ouvrier("AHMED","KAMAL",100,1990,2006)
ouvrier_2 = Ouvrier("ALi","GAMAL",101,1980,2016)
ouvrier_3 = Ouvrier("LACHGAR","YASSIN",102,1989,2025)

cadre_1 = Cadre("AITMA","FATIM",105,2000,1)
cadre_2 = Cadre("RAHA","KHALID",106,1999,2)
cadre_3 = Cadre("FILALI","MOHAMED",106,2003,3)
cadre_4 = Cadre("SAID","TARIQ",106,2003,4)

Patron.chiffre_affaire = 1_000_000
patron_1 = Patron("LAFSIOUI",99,"HAMZA",2001)

# =========================================
# TESTING 
patron_1.chiffre_affaire = 10
print(patron_1.chiffre_affaire)
print(Patron.chiffre_affaire)
# =========================================
employees = [ouvrier_1,ouvrier_2,ouvrier_3,cadre_1,cadre_2,cadre_3,patron_1]

for emp in employees:
    print(emp)
    print("="*120)

try:
    print(cadre_4)
except Exception as e:
    print(e)
```

### Challenge 04: la gestion des employés.

Travail demandé:

L'IR (impôt sur le revenu) est un impôt global établi sur la totalité des revenus (le salaire) dont dispose une personne au cours d'une année déterminée. Pour déterminer le montant de l'IR on utilise le barème suivant :

| **Tranches de salaire annuel (en DH)** | **Taux IR en %** |
| -------------------------------------- | ---------------- |
| 0 à 28 000                             | 0 %              |
| 28 001 à 40 000                        | 12 %             |
| 40 001 à 50 000                        | 24 %             |
| 50 001 à 60 000                        | 34 %             |
| 60 001 à 150 000                       | 38 %             |
| Au-delà de 150 000                     | 40 %             |

Exemple : si le salaire d'un employé est 9 500 DH, alors son salaire annuel = salaire × 12 = 9 500 × 12 = 114 000 DH, donc le taux IR = 38 %.

1. Créer la classe IR avec :

a. L'attribut _tranches : un tableau statique (attribut de classe) initialisé par les valeurs des tranches ci-dessus.

b. L'attribut _taux_ir : un tableau statique de taux IR initialisé par les valeurs correspondantes.

c. Une méthode statique get_ir(salaire) qui retourne le taux d'IR correspondant au salaire passé en paramètre.

2. Créer l'interface (classe abstraite) IEmploye avec les méthodes suivantes :

a. Une méthode age() qui retournera l'âge d'un employé (entier).

b. Une méthode anciennete() qui retournera l'ancienneté d'un employé (le nombre d'années travaillées).

c. Une méthode date_retraite(age_retraite) qui renvoie la date de retraite : date de retraite = date de naissance + âge de retraite.

Un employé est caractérisé par : Matricule (mtle), Nom (nom), Date de naissance (date_naissance), Date d'embauche (date_embauche), Salaire de base (salaire_base).

e. L'implémentation de l'interface IEmploye avec ces trois méthodes.

f. La redéfinition de la méthode **__str__**() qui renvoie toutes les propriétés.

Un formateur est un employé avec en plus : le nombre d'heures supplémentaires par mois (heure_sup) et la rémunération par heure supplémentaire, dont la valeur est partagée par tous les formateurs et par défaut égale à 70,00 DH (remuneration_hsup).

4. Créer la classe Formateur avec :

a. Les attributs.

b. L'accesseur remuneration_hsup.

c. Les deux constructeurs (par défaut et d'initialisation).

d. La méthode salaire_a_payer() pour calculer le salaire net d'un formateur sachant que :

Salaire net = (salaire_base + heure_sup × remuneration_hsup) × (1 − taux IR)

N.B. : Utiliser la méthode get_ir de la classe IR pour calculer le taux IR.


```python
from abc import ABC, abstractmethod
from datetime import date


class IR:

    # Attributs de classe
    _tranches = [28000, 40000, 50000, 60000, 150000]
    _taux_ir = [0, 0.12, 0.24, 0.34, 0.38, 0.40]

    @staticmethod
    def get_ir(salaire):
        if salaire <= 28000:
            return IR._taux_ir[0]

        elif salaire <= 40000:
            return IR._taux_ir[1]

        elif salaire <= 50000:
            return IR._taux_ir[2]

        elif salaire <= 60000:
            return IR._taux_ir[3]

        elif salaire <= 150000:
            return IR._taux_ir[4]

        else:
            return IR._taux_ir[5]
    


class IEmploye(ABC):


    @abstractmethod
    def age(self):
        pass

    @abstractmethod
    def anciennete(self):
        pass

    @abstractmethod
    def date_retraite(self, age_retraite):
        pass




class Employe(IEmploye, ABC):

    def __init__(
        self,
        mtle=None,
        nom=None,
        date_naissance=None,
        date_embauche=None,
        salaire_base=0
    ):
        self._mtle = mtle
        self._nom = nom
        self.date_naissance = date_naissance
        self.date_embauche = date_embauche
        self._salaire_base = salaire_base

  
    @property
    def date_naissance(self):
        return self._date_naissance

    @date_naissance.setter
    def date_naissance(self, value):
        self._date_naissance = value

   
    @property
    def date_embauche(self):
        return self._date_embauche

    @date_embauche.setter
    def date_embauche(self, value):
        self._date_embauche = value


    def age(self):
        today = date.today()

        age = today.year - self.date_naissance.year

        if (today.month, today.day) < (
            self.date_naissance.month,
            self.date_naissance.day
        ):
            age -= 1

        return age

 
    def anciennete(self):
        today = date.today()

        anciennete = today.year - self.date_embauche.year

        if (today.month, today.day) < (
            self.date_embauche.month,
            self.date_embauche.day
        ):
            anciennete -= 1

        return anciennete

   
    def date_retraite(self, age_retraite):
        return self.date_naissance.replace(
            year=self.date_naissance.year + age_retraite
        )

   
    @abstractmethod
    def salaire_a_payer(self):
        pass

    
    def __str__(self):
        return (
            f"Matricule: {self._mtle} | "
            f"Nom: {self._nom} | "
            f"Date de naissance: {self.date_naissance} | "
            f"Date d'embauche: {self.date_embauche} | "
            f"Salaire de base: {self._salaire_base}"
        )
    


class Formateur(Employe):


    _remuneration_hsup = 70.00

    def __init__(
        self,
        mtle=None,
        nom=None,
        date_naissance=None,
        date_embauche=None,
        salaire_base=0,
        heure_sup=0
    ):
        super().__init__(
            mtle,
            nom,
            date_naissance,
            date_embauche,
            salaire_base
        )

        self._heure_sup = heure_sup

  
    @property
    def remuneration_hsup(self):
        return Formateur._remuneration_hsup

    @remuneration_hsup.setter
    def remuneration_hsup(self, value):
        Formateur._remuneration_hsup = value

 
    def salaire_a_payer(self):

        # Salaire annuel
        salaire_annuel = self._salaire_base * 12

        # Taux IR
        taux_ir = IR.get_ir(salaire_annuel)

        # Salaire net mensuel    
        salaire_net = (
            self._salaire_base
            + self._heure_sup * self.remuneration_hsup
        ) * (1 - taux_ir)

        return salaire_net

    
    def __str__(self):
        return (
            super().__str__()
            + f" | Heures supplementaires: {self._heure_sup}"
            + f" | Remuneration heure sup: {self.remuneration_hsup}"
            + f" | Salaire a payer: {self.salaire_a_payer()}"
        )
    


f = Formateur(101,"HAMZA",date(2001, 5, 10),date(2022, 9, 1),9500,10)

print(f)

print("Age :", f.age())

print("Anciennete :", f.anciennete())

print("Date retraite :", f.date_retraite(60))

print("Taux IR :", IR.get_ir(f._salaire_base * 12))

print("Salaire a payer :", f.salaire_a_payer())
```
